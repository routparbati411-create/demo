**KKNImputer**

- sklearn

  - impute

     - KNNImputer

       - fit_transform

- it is a distance based method

- neighbour means closet distance once 

- the user will provide i

- instead of taking avg of all the values , first will distance between data points

- which are closet will take those value avg only

```python
data=np.array([[25,50000],[26,52000],[30,60000],[28,np.nan]])
pd.DataFrame(data,columns=['Age','Salary'])

# i want fill the row [28,np.nan]
# look at Age column not income column
# for 28 which values are close: pick only 2 closest points i.e. neighbours
# 26 and 30
# for 26 what is income : 50000
# for 30 what is income : 60000
# take the avg of (52000+60000)/2=56000
```

```python
from sklearn.impute import KNNImputer
imputer=KNNImputer(n_neighbors=2)
imputer.fit_transform(data)
```
