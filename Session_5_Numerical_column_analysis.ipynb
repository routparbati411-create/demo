**pd.crosstab**

- Crosstab is used in EDA to analyze the relationship between categorical variables by creating a frequency table. 
- It helps identify patterns, compare categories, understand distributions, and perform feature analysis before model building.
- It is especially useful for studying how one categorical variable affects another.

```python
- dtypes: Used to check the data type of each column in a DataFrame.

- select_dtypes : Used to select columns based on their data types. categorial and numerical column

- keys() : Returns column names of a DataFrame.

- values() : Returns the DataFrame data as a NumPy array.

- columns() : Returns column names.

- index() : Returns row labels.

- shape() : Returns number of rows and columns.

- size() : Returns total number of elements.

- ndim() : Returns number of dimensions.

- info() : Provides complete information about the dataset.

- describe() : Provides statistical summary like mean,median,std,count,min,max,25,50,75 % 

- umique() : Returns unique values.

- nunique() : Counts unique values.

- value_counts() : Counts occurrences of each value.

- isnull().sum() : Check missing values.

- sample() : Random rows.

- corr() : Correlation between numerical columns.

- groupby() : Group and aggregate data.

- sort_values() : Sort data.

- drop() : Remove rows/columns.

- rename() : Rename columns.
```
