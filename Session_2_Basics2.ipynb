**task**

- separate categorial and numerical columns in different list

- using dtype 

```python
# idea
# extract keys and values
# apply the for loop
# check the condition if the data type = object then
# append in cat list
# otherwise append in num list

keys=visa_df.dtypes.keys()
values=visa_df.dtypes.values
cat=[]
num=[]
for i,j in zip(keys,values):
    if j=='object':
        cat.append(i)
    else:
        num.append(i)
cat,num
```

```python
keys=visa_df.dtypes.keys()
values=visa_df.dtypes.values
cat=[]
num=[]
for i,j in zip(keys,values):
    if j=='object':
        cat.append(j)
    else:
        num.append(j)
cat,num
```

```python
keys=visa_df.dtypes.keys()
values=visa_df.dtypes.values
cat=[]
num=[]
for i,j in zip(keys,values):
    if j==object:
        cat.append([i,j])
    else:
        num.append([i,j])
cat,num
```

```python
keys=visa_df.dtypes.keys()
values=visa_df.dtypes.values
cat=[]
num=[]
for i,j in zip(keys,values):
    if j==object:
        cat.extend([i,j])
    else:
        num.extend([i,j])
cat,num
```

```python
data_types=visa_df.dtypes
cat=[]
num=[]
for i in data_types.keys():
    if data_types[i]=='object':
        cat.append([i,data_types[i]])
    else:
        num.append([i,data_types[i]])    
        
cat,num
print(*cat,sep='\n')
print()
print(*num,sep='\n')
```
