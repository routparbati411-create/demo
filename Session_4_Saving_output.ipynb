**Pie_Chart**

```python
import os
# step-1
current_dir=os.getcwd()
current_dir
# step-2
new_folder="Pie_Chart"
os.makedirs(new_folder,exist_ok=True)
# step-3:
new_dir=os.path.join(current_dir,new_folder)
new_dir
# step-4
for i,j in enumerate(cat[1:]):
    column_path=f'{j}.jpg'
    full_path=os.path.join(new_dir,column_path)
    keys=visa_df[j].value_counts().keys()
    values=visa_df[j].value_counts().values
    plt.figure(figsize=(7,8))
    plt.subplot(3,3,i+1)
    plt.pie(values,labels=keys,autopct='%1.1f%%')
    plt.title(f'{j}_pie_chart')
    plt.savefig(full_path)
    plt.show()
```
