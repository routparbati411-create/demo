**One Hot Encoder**

- One hot means at a time only one value **ON**

- Other will be **OFF**

- ON means one value become **True**

- OFF means one value become **False**

- True means one value become **1**

- False means one value become **0**

- for example : **case status** has two unique levels
    - Certified
    
    - Denied

- So one hot encoder creates two new columns

   - case_status_certified

   - case_status_denied
    

|Case Status | Certified | Denied |
|------------|-----------|--------|
|Certified   |       1   |       0|
|Denied      |0          |1       |
|Certified   |       1   |       0|
|Denied      |0          |1       |
